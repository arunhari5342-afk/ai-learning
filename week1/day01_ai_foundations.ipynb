## 1. AI Foundations

### What is Artificial Intelligence?

Artificial Intelligence (AI) is the field of building computer systems that can perform tasks that normally require human intelligence.

Examples include:

- Image recognition
- Speech recognition
- Recommendation systems
- Spam detection
- Language understanding
- Generative AI

### Machine Learning

Machine Learning (ML) is a subset of AI where computers learn patterns from data instead of being explicitly programmed with every rule.

A simple Machine Learning workflow is:

**Data → Preprocess → Train → Evaluate → Infer**

### Narrow AI

Narrow AI is designed to perform a specific task or a limited set of tasks.

Examples:

- Spam detection
- Face recognition
- Recommendation systems
- Speech recognition

### General AI

Artificial General Intelligence (AGI) refers to the concept of an AI system capable of performing a broad range of intellectual tasks rather than being restricted to one specific task.

### AI → ML → Deep Learning → LLMs / GenAI

A useful mental model is:

**AI**
→ **Machine Learning**
→ **Deep Learning**
→ **Transformers**
→ **LLMs**

Generative AI includes systems that generate text, images, audio, video, and code.

## 2. Three Machine Learning Paradigms

### Supervised Learning

Supervised learning uses training data containing inputs and known correct outputs.

Example:

**House features → House price**

Other examples:

- Spam vs. non-spam classification
- Predicting house prices
- Fraud detection

### Unsupervised Learning

Unsupervised learning works with data without predefined labels and attempts to discover patterns or structure.

Example:

**Customer data → Customer groups**

A common technique is clustering.

Examples:

- Customer segmentation
- Grouping similar documents
- Finding patterns in data

### Reinforcement Learning

Reinforcement learning involves an agent interacting with an environment and learning from rewards or penalties.

Example:

**Game-playing agent → Action → Reward → Learning**

Examples:

- Game-playing systems
- Robotics
- Control systems

### Quick comparison

| Paradigm | Learning signal | Example |
|---|---|---|
| Supervised | Correct labels | Spam detection |
| Unsupervised | Hidden structure | Customer clustering |
| Reinforcement | Rewards / penalties | Game playing |

## 4. Machine Learning Lifecycle

A typical Machine Learning workflow is:

**Data → Preprocess → Train → Evaluate → Infer**

### 1. Data

Collect the information needed for the problem.

Example:

Student information containing age, department, and score.

### 2. Preprocess

Clean and transform the data.

Common preprocessing tasks include:

- Handling missing values
- Removing duplicates
- Encoding categorical values
- Scaling numerical values

### 3. Train

Give training data to a Machine Learning algorithm so that it can learn patterns.

### 4. Evaluate

Test the trained model using data that it has not seen during training.

Common evaluation metrics include:

- Accuracy
- Precision
- Recall
- F1 score
- MAE
- MSE

### 5. Infer

Use the trained model to make predictions on new data.

Example:

**New house information → Trained model → Predicted house price**

## 5. NumPy

NumPy is a Python library used for numerical computing.

It provides efficient arrays and mathematical operations that are widely used in Machine Learning and Data Science.

### NumPy Array

A NumPy array is a collection of values stored in a structured numerical format.

A one-dimensional NumPy array can be used to represent a vector.

Example:

[10, 20, 30, 40]

````python
import numpy as np
````

````python
numbers = np.array([10, 20, 30, 40])

print(numbers)
print(type(numbers))
````

### Shape

The shape of a NumPy array tells us its dimensions.

For a one-dimensional array containing four values:

(4,)

For a two-dimensional array containing three rows and two columns:

(3, 2)

The shape is very important in Machine Learning because models work with structured numerical data.

````python
numbers = np.array([10, 20, 30, 40])

print("Array:", numbers)
print("Shape:", numbers.shape)
````

````python
matrix = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])

print(matrix)
print("Shape:", matrix.shape)
````

### dtype

dtype means "data type".

NumPy arrays store values using a specific data type.

Common numerical data types include:

- int
- float
- bool

The dtype tells us what kind of values are stored inside the NumPy array.

````python
integers = np.array([10, 20, 30])

decimals = np.array([10.5, 20.5, 30.5])

print("Integer dtype:", integers.dtype)
print("Decimal dtype:", decimals.dtype)
````

### Broadcasting

Broadcasting allows NumPy to perform operations between arrays and individual values without manually repeating the value.

For example:

[1, 2, 3] + 10

becomes:

[11, 12, 13]

NumPy automatically applies 10 to every element of the array.

Broadcasting is very important when performing numerical operations efficiently.

````python
numbers = np.array([1, 2, 3])

result = numbers + 10

print("Original:", numbers)
print("Result:", result)
````

### Dot Product

The dot product combines corresponding values from two vectors and adds the results.

For example:

A = [1, 2, 3]

B = [4, 5, 6]

The dot product is:

1 × 4 + 2 × 5 + 3 × 6 = 32

Dot products are fundamental to Machine Learning and neural networks.

````python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

result = np.dot(a, b)

print("Vector A:", a)
print("Vector B:", b)
print("Dot product:", result)
````

````python
result = a @ b

print(result)
````

## 6. Pandas

Pandas is a Python library used for working with structured data.

A Pandas DataFrame is similar to a table with rows and columns.

Pandas is commonly used for:

- Reading datasets
- Exploring data
- Filtering data
- Grouping data
- Cleaning and preparing data

### Reading a CSV file

Pandas provides `read_csv()` to load data from a CSV file into a DataFrame.

````python
import pandas as pd
````

````python
df = pd.read_csv("../datasets/students.csv")

df
````

The `head()` method displays the first rows of a DataFrame.

It is useful for quickly checking whether a dataset was loaded correctly.

By default, `head()` displays the first five rows.

````python
df.head()
````

````python
df.head(3)
````

The `info()` method provides structural information about a DataFrame.

The `describe()` method provides statistical information about numerical columns.

````python
df.info()
````

### Filtering

Filtering allows us to select rows that satisfy a condition.

For example, we can find students whose score is greater than 80.

````python
high_scores = df[df["score"] > 80]

high_scores
````

````python
computer_science = df[df["department"] == "Computer Science"]

computer_science
````

### `groupby()`

The `groupby()` method allows us to divide data into groups and calculate statistics for each group.

For example, we can calculate the average score for each department.

````python
department_scores = df.groupby("department")["score"].mean()

department_scores
````

````python
department_counts = df.groupby("department")["name"].count()

department_counts
````

## 7. Matplotlib

Matplotlib is a Python library used for data visualization.

Visualization helps us understand patterns and relationships in data.

Today I will create two types of charts:

1. Histogram
2. Bar chart

````python
import matplotlib.pyplot as plt
````

### Histogram

A histogram shows the distribution of numerical values.

It divides numerical values into ranges called bins and shows how many values fall into each range.

Here I will use the student scores to see how the scores are distributed.

````python
plt.hist(df["score"], bins=5)

plt.title("Distribution of Student Scores")
plt.xlabel("Score")
plt.ylabel("Number of Students")

plt.show()
````

### Bar Chart

A bar chart is useful for comparing different categories.

Here I will compare the average score of each department.

The departments are categorical data, while the average score is numerical data.

````python
department_scores = df.groupby("department")["score"].mean()

department_scores.plot(kind="bar")

plt.title("Average Score by Department")
plt.xlabel("Department")
plt.ylabel("Average Score")

plt.show()
````
