````python
import nltk
import pandas as pd

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer

from preprocess import preprocess_text
from preprocess import preprocess_to_string
````

## 1. NLTK Tokenization

Tokenization is the process of breaking text into smaller units called tokens.

For example:

```text
"Natural language processing is powerful."

````python
text = """
Natural language processing is powerful.
It helps computers understand human language.
"""

tokens = word_tokenize(text)

print(tokens)
````

## 2. Remove Stop Words

Stop words are common words that often provide limited information for NLP tasks.

Examples include:

- the
- is
- a
- an
- and
- to
- of
- in

We will remove these words from our tokens.

````python
stop_words = set(stopwords.words("english"))

filtered_tokens = [
    token
    for token in tokens
    if token.lower() not in stop_words
]

print(filtered_tokens)
````

## 3. Lemmatization with NLTK

Lemmatization converts words into their base or dictionary form.

Examples:

- cars → car
- studies → study
- running → running

We will use NLTK's WordNetLemmatizer to understand the basic idea of lemmatization.

````python
lemmatizer = WordNetLemmatizer()

lemmatized_tokens = [
    lemmatizer.lemmatize(token.lower())
    for token in filtered_tokens
]

print(lemmatized_tokens)
````

## 4. Text Preprocessing with spaCy

Now we will use our reusable `preprocess.py` utility.

The `preprocess_text()` function performs:

1. Lowercasing
2. Tokenization
3. Stop-word removal
4. Punctuation removal
5. Number removal
6. Lemmatization

This allows us to reuse the same preprocessing logic in other NLP projects.

````python
text = """
Natural language processing allows computers to understand
human languages. Computers are learning from large datasets.
"""

cleaned_tokens = preprocess_text(text)

print(cleaned_tokens)
````

## 5. Convert Preprocessed Tokens into Text

Machine learning vectorizers expect text documents.

Our `preprocess_to_string()` function converts the cleaned tokens back into one clean text string.

Example:

Raw text:

"Computers are learning from datasets."

Cleaned text:

"computer learn dataset"

````python
cleaned_text = preprocess_to_string(text)

print(cleaned_text)
````

## 6. Create Sample Documents

We need multiple documents to demonstrate CountVectorizer and TF-IDF.

Each document represents one text example.

Our documents will contain information about:

- Natural Language Processing
- Machine Learning
- Deep Learning
- Chatbots

````python
documents = [
    """
    Natural language processing helps computers understand human language.
    NLP is used in chatbots, search engines, and virtual assistants.
    """,

    """
    Machine learning allows computers to learn from data.
    Machine learning is widely used for prediction and classification.
    """,

    """
    Deep learning uses neural networks to learn complex patterns.
    Neural networks are useful for images, speech, and language processing.
    """,

    """
    Chatbots use natural language processing to understand user questions.
    Modern chatbots can generate useful responses using machine learning.
    """
]

print(f"Number of documents: {len(documents)}")
````

## 7. Preprocess All Documents

We will apply our reusable preprocessing function to every document.

The result will be clean text that can be passed to the scikit-learn vectorizers.

````python
cleaned_documents = [
    preprocess_to_string(document)
    for document in documents
]

for index, document in enumerate(cleaned_documents, start=1):
    print(f"Document {index}:")
    print(document)
    print()
````

## 8. CountVectorizer

`CountVectorizer` converts text documents into numerical vectors based on word frequency.

For example:

"machine learning machine"

becomes:

machine → 2

learning → 1

Each unique word becomes a feature, and the value represents how many times that word appears.

````python
count_vectorizer = CountVectorizer()

count_matrix = count_vectorizer.fit_transform(
    cleaned_documents
)

print(count_matrix)
````

## 9. Inspect CountVectorizer Vocabulary

The vectorizer creates a vocabulary containing the unique words found in our documents.

Each vocabulary word becomes a feature or column in the numerical representation.

````python
vocabulary = count_vectorizer.get_feature_names_out()

print(vocabulary)
````

## 10. CountVectorizer Matrix

Each row represents a document.

Each column represents a vocabulary word.

Each value represents the number of times that word appears in the document.

````python
count_df = pd.DataFrame(
    count_matrix.toarray(),
    columns=count_vectorizer.get_feature_names_out()
)

count_df
````

## 11. TfidfVectorizer

TF-IDF stands for:

**Term Frequency — Inverse Document Frequency**

TF-IDF gives higher importance to words that are:

- frequent in a particular document
- relatively rare across the collection of documents

Words appearing in many documents generally receive lower importance.

### Main idea

Frequent in one document + Rare across documents = Higher TF-IDF score

````python
tfidf_vectorizer = TfidfVectorizer()

tfidf_matrix = tfidf_vectorizer.fit_transform(
    cleaned_documents
)

print(tfidf_matrix)
````

## 12. Inspect the TF-IDF Matrix

The TF-IDF vectorizer converts each document into a numerical vector.

Each value represents the importance of a particular word in that document.

A higher TF-IDF value generally means that the word is more useful for distinguishing that document from the others.

````python
tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf_vectorizer.get_feature_names_out()
)

tfidf_df
````

## 13. Inspect Top TF-IDF Terms

We will identify the words with the highest TF-IDF scores for each document.

This helps us understand which words are most important or representative of each document.

````python
def get_top_tfidf_terms(
    tfidf_matrix,
    vectorizer,
    top_n=5
):
    feature_names = vectorizer.get_feature_names_out()

    results = []

    for document_index in range(tfidf_matrix.shape[0]):

        scores = tfidf_matrix[
            document_index
        ].toarray().flatten()

        top_indices = scores.argsort()[::-1][:top_n]

        terms = [
            (
                feature_names[index],
                scores[index]
            )
            for index in top_indices
            if scores[index] > 0
        ]

        results.append(terms)

    return results
````

````python
top_terms = get_top_tfidf_terms(
    tfidf_matrix,
    tfidf_vectorizer,
    top_n=5
)

for document_number, terms in enumerate(
    top_terms,
    start=1
):

    print(f"\nDocument {document_number}")

    for term, score in terms:
        print(f"{term:20} {score:.4f}")
````

## 14. Reusable Text → Vector Function

In a real NLP or machine learning application, we need to convert new text into the same numerical representation used during training.

The process is:

Raw Text
↓
Preprocessing
↓
Clean Text
↓
Fitted Vectorizer
↓
Numerical Vector

The vectorizer must already be fitted on the training documents.

For new text, we use:

`vectorizer.transform()`

We should not fit the vectorizer again on new text.

````python
def text_to_vector(
    text: str,
    vectorizer
):
    """
    Convert raw text into a numerical vector.

    The vectorizer must already be fitted.
    """

    cleaned_text = preprocess_to_string(text)

    vector = vectorizer.transform(
        [cleaned_text]
    )

    return vector
````

## 15. Test the Text → Vector Function

We will provide a new sentence to our reusable function.

The new sentence will:

1. Be preprocessed.
2. Be converted using the vocabulary learned by the vectorizer.
3. Be transformed into a numerical TF-IDF vector.

The vectorizer will not be fitted again.

````python
new_text = """
Natural language processing is used to build intelligent chatbots.
"""

new_vector = text_to_vector(
    new_text,
    tfidf_vectorizer
)

print(new_vector.toarray())
````

## 16. Inspect the New Text Vector

The new text vector has the same number of features as the vocabulary learned by the TF-IDF vectorizer.

This is important because machine learning models require training data and new data to use the same feature structure.

````python
print("Vector shape:", new_vector.shape)

print(
    "Number of features:",
    len(tfidf_vectorizer.get_feature_names_out())
)
````

## 17. Training Data vs New Data

For training data, we use:

`vectorizer.fit_transform(training_documents)`

This learns the vocabulary and transforms the documents.

For new data, we use:

`vectorizer.transform(new_documents)`

We do not fit the vectorizer again on new data.

This ensures that both training data and new data use the same vocabulary and feature space.

````python
training_vectors = tfidf_vectorizer.fit_transform(
    cleaned_documents
)

new_text_vector = tfidf_vectorizer.transform(
    [preprocess_to_string(new_text)]
)

print("Training vector shape:", training_vectors.shape)
print("New text vector shape:", new_text_vector.shape)
````
