## 4. Text Summarization

Text summarization creates a shorter version of a longer piece of text
while attempting to preserve the important information.

We will use a pretrained Transformer summarization model.

```python
from transformers import pipeline

text_generator = pipeline(
    "text-generation",
    model="distilgpt2"
)

print("Text generation pipeline loaded successfully")
```

```python
article = """
Artificial intelligence is changing many industries around the world.
Machine learning allows computers to learn patterns from data without
being explicitly programmed for every individual task. Natural language
processing focuses on enabling computers to understand and work with
human language. Modern Transformer models have significantly improved
many NLP applications including translation, summarization, question
answering, and text generation.
"""

prompt = f"""
Summarize this article briefly:

{article}

Summary:
"""

result = text_generator(
    prompt,
    max_new_tokens=60,
    do_sample=False
)

print("Generated text:")
print(result[0]["generated_text"])
```

### Summarization Observation

The text-generation model generates text based on the provided prompt.

However, DistilGPT2 is a general text-generation model rather than a
model specifically fine-tuned for summarization.

This demonstrates that choosing an appropriate pretrained model for a
specific NLP task is important.

## 5. Zero-Shot Classification

Zero-shot classification allows us to classify text into categories
provided at inference time.

The model can compare the input text with candidate labels without
requiring us to train a new classifier for those labels.

```python
from transformers import pipeline

classifier = pipeline(
    "zero-shot-classification",
    model="valhalla/distilbart-mnli-12-1"
)

print("Zero-shot classification pipeline loaded successfully")
```

```python
text = "I am learning Python, machine learning, and neural networks."

candidate_labels = [
    "technology",
    "sports",
    "politics",
    "cooking"
]

result = classifier(
    text,
    candidate_labels
)

print("Text:")
print(text)

print("\nLabels:")
print(result["labels"])

print("\nScores:")
print(result["scores"])
```

### Zero-Shot Observation

The model receives:

1. The input text
2. A list of candidate categories

It then ranks the categories according to how well they match the text.

For this example, the `technology` category should generally receive a
high score because the text discusses Python, machine learning, and
neural networks.

```python
examples = [
    "The football team won the championship.",
    "Python is widely used in machine learning.",
    "The government announced a new policy."
]

candidate_labels = [
    "sports",
    "technology",
    "politics"
]

for text in examples:
    result = classifier(
        text,
        candidate_labels
    )

    print("Text:", text)
    print("Prediction:", result["labels"][0])
    print("Score:", result["scores"][0])
    print()
```

## 6. Tokenization

Before text is processed by a Transformer, raw text must be converted
into tokens.

A tokenizer converts text into tokens and then maps those tokens to
numerical token IDs that the model can process.

```python
from transformers import AutoTokenizer

print("AutoTokenizer imported successfully")
```

```python
tokenizer = AutoTokenizer.from_pretrained(
    "bert-base-uncased"
)

print("Tokenizer loaded successfully")
```

```python
text = "I love learning NLP!"

tokens = tokenizer.tokenize(text)

print("Original text:")
print(text)

print("\nTokens:")
print(tokens)
```

### Understanding Subword Tokenization

A tokenizer does not always represent an entire word as one token.

Some words can be split into smaller subword pieces.

This allows Transformer models to handle uncommon words and words that
may not appear as complete vocabulary entries.

## 7. Inspect Tokenization on Tricky Text

We will test punctuation, contractions, hyphenated words, and uncommon
words to observe how the tokenizer handles them.

```python
tricky_text = """
Transformers are amazing!
I can't believe NLP models understand words like unbelievable,
AI-powered, and Chennai-based.
"""

tokens = tokenizer.tokenize(tricky_text)

print("Original text:")
print(tricky_text)

print("\nTokens:")
print(tokens)
```

```python
encoded = tokenizer(tricky_text)

print("Encoded output:")
print(encoded)
```

### Understanding the Encoded Output

The tokenizer output may contain:

- `input_ids` — numerical IDs representing tokens
- `attention_mask` — identifies valid input positions
- `token_type_ids` — identifies input segments for models that use them

The exact fields depend on the tokenizer and model.

```python
print("Input IDs:")
print(encoded["input_ids"])

print("\nAttention Mask:")
print(encoded["attention_mask"])
```

```python
decoded_tokens = tokenizer.convert_ids_to_tokens(
    encoded["input_ids"]
)

print("Tokens reconstructed from IDs:")
print(decoded_tokens)
```

```python
decoded_text = tokenizer.decode(
    encoded["input_ids"]
)

print("Decoded text:")
print(decoded_text)
```

## 8. Compare Two Pretrained Models

Different pretrained models can produce different predictions for the
same input.

We will compare two sentiment-analysis models using exactly the same
sentence.

```python
from transformers import pipeline

print("Pipeline imported successfully")
```

```python
sentiment_model_1 = pipeline(
    "sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)

print("Model 1 loaded successfully")
```

```python
sentiment_model_2 = pipeline(
    "text-classification",
    model="textattack/bert-base-uncased-SST-2"
)

print("Model 2 loaded successfully")
```

```python
text = "The movie was surprisingly good, although the ending was disappointing."

result_1 = sentiment_model_1(text)
result_2 = sentiment_model_2(text)

print("Input:")
print(text)

print("\nModel 1:")
print(result_1)

print("\nModel 2:")
print(result_2)
```

## 8. Model Comparison Observation

Both pretrained models received exactly the same input sentence.

Model 1 predicted the sentiment as POSITIVE with a confidence score
of approximately 99.97%.

Model 2 produced LABEL_1 with a confidence score of approximately
96.03%.

The models produced different label names because different pretrained
models may use different label mappings. Therefore, label names such as
LABEL_0 and LABEL_1 should be checked using the model configuration
before interpreting them as positive or negative.

The confidence scores are also different because the models were trained
with different architectures, parameters, and training procedures.

This demonstrates that different pretrained Transformer models can
produce different outputs for the same input.
