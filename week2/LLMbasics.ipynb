```python
import os
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv()
```

```python

api_key = os.getenv("OPENAI_API_KEY")

print("API key loaded:", api_key is not None)
```

```python
from openai import OpenAI

client = OpenAI(
    base_url="https://api.groq.com/openai/v1",
    api_key=api_key
)

print("LLM client created")
```

```python
response = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=[
        {
            "role": "user",
            "content": "Write a short paragraph explaining artificial intelligence."
        }
    ]
)

print(response.choices[0].message.content)
```

```python
zero_shot_prompt = """
Classify the sentiment of this sentence as Positive, Negative, or Neutral.

Sentence:
"The product works perfectly."
"""

response = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=[
        {"role": "user", "content": zero_shot_prompt}
    ],
    temperature=0
)

print(response.choices[0].message.content)
```

```python
few_shot_prompt = """
Classify the sentiment as Positive, Negative, or Neutral.

Example 1:
Sentence: "I love this product."
Answer: Positive

Example 2:
Sentence: "The product stopped working."
Answer: Negative

Example 3:
Sentence: "The package arrived yesterday."
Answer: Neutral

Now classify:

Sentence:
"The product works perfectly."

Answer:
"""

response = client.chat.completions.create(
     model="openai/gpt-oss-20b",
    messages=[
        {"role": "user", "content": few_shot_prompt}
    ],
    temperature=0
)

print(response.choices[0].message.content)
```

```python
text = """
John Smith works as a software engineer at ABC Technologies.
He has 5 years of experience and specializes in Python, FastAPI,
and PostgreSQL.
"""
```

```python
extraction_prompt = f"""
Extract the following information from the text:

- person_name
- job_title
- company
- experience_years
- skills

Text:
{text}

Return the result as JSON.
"""
```

```python
response = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=[
        {"role": "user", "content": extraction_prompt}
    ],
    temperature=0
)

print(response.choices[0].message.content)
```

```python
print(response.usage)
```

```python
usage = response.usage

print("Input tokens :", usage.prompt_tokens)
print("Output tokens:", usage.completion_tokens)
print("Total tokens :", usage.total_tokens)
```
